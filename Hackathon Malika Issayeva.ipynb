# Maps

```python
#Для карты по смертям и их причин, нам нужен csv файл с широтой и долготой всех штатов сша
usa = pd.read_csv('C:/Users/malik/Documents/Python samsung/hackathon/us-state-capitals.csv')
usa.head()
```

```python
usa.rename(columns = {'name':'state'},inplace=True)
usa = usa[['state', 'latitude', 'longitude']]
usa.head()
```

```python
#Задаем широту и долготу, а так же нужную нам информацию
state = death
state1 = state.copy()
State = state1['State']
Cause = state1['CauseName']
deaths = state1['Deaths']
lon = usa.longitude.value_counts().index
lat = usa.latitude.value_counts().index
```

```python
#создаем карту с помощью folium, используем HTML для маркеров
m = folium.Map(location = [48, -102],
               zoom_start = 3,
               tiles = 'OpenStreetMap')
#добавляем различные стили карт
folium.raster_layers.TileLayer('Open Street Map').add_to(m)
folium.raster_layers.TileLayer('Stamen Terrain').add_to(m)
folium.raster_layers.TileLayer('Stamen Toner').add_to(m)
folium.raster_layers.TileLayer('Stamen Watercolor').add_to(m)
folium.raster_layers.TileLayer('CartoDB Dark_Matter').add_to(m)
folium.raster_layers.TileLayer('CartoDB Positron').add_to(m)

folium.LayerControl().add_to(m)

for i in range(50):
    info = '<b>State:</b>' + str(State[i]) + '<br><b>Cause</b>: ' + str(Cause[i]) + '<br><b>Deaths</b>: ' + str(deaths[i])
    iframe = folium.IFrame(html = info, width = 200, height = 100)
    folium.Marker(location = [lat[i], lon[i]],
                  popup = folium.Popup(iframe),
                  icon = folium.Icon(color = 'green')
                 ).add_to(m)
    
m
```
